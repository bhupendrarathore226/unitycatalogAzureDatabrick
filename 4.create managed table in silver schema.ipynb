```python
import logging
import json

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s - %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)
logger = logging.getLogger("create_managed_table_silver")
logger.info("Notebook started: create managed table in silver schema")
logger.info("Starting: Drop and recreate silver managed table: drivers")
```

```python
try:
    spark.sql("DROP TABLE IF EXISTS finance_dev.silver_dev.drivers")
    spark.sql("""
        CREATE TABLE IF NOT EXISTS finance_dev.silver_dev.drivers
        AS
        SELECT driverId     AS driver_id,
               driverRef    AS driver_ref,
               number,
               code,
               concat(name.forename, ' ', name.surname) AS name,
               nationality,
               dob,
               current_timestamp() AS ingestion_date
        FROM finance_dev.bronze_dev.drivers
    """)
    logger.info("Completed: Silver table finance_dev.silver_dev.drivers created successfully")
except Exception as e:
    logger.error("Failed to create silver table: drivers. Error: %s", str(e), exc_info=True)
    dbutils.notebook.exit(json.dumps({"status": "FAILED", "step": "create_silver_drivers_table", "error": str(e)}))
    raise
```

```python
logger.info("Starting: Drop and recreate silver managed table: results")
```

```python
try:
    spark.sql("DROP TABLE IF EXISTS finance_dev.silver_dev.results")
    spark.sql("""
        CREATE TABLE IF NOT EXISTS finance_dev.silver_dev.results
        AS
        SELECT resultId          AS result_id,
               raceId            AS race_id,
               driverId          AS driver_id,
               constructorId     AS constructor_id,
               number,
               grid,
               position,
               positionText      AS position_text,
               positionOrder     AS position_order,
               points,
               laps,
               time,
               milliseconds,
               fastestLap,
               rank,
               fastestLapTime    AS fastest_lap_time,
               fastestLapSpeed   AS fastest_lap_speed,
               statusId          AS status_id,
               current_timestamp() AS ingestion_date
        FROM finance_dev.bronze_dev.results
    """)
    logger.info("Completed: Silver table finance_dev.silver_dev.results created successfully")
except Exception as e:
    logger.error("Failed to create silver table: results. Error: %s", str(e), exc_info=True)
    dbutils.notebook.exit(json.dumps({"status": "FAILED", "step": "create_silver_results_table", "error": str(e)}))
    raise
```

```python
logger.info("Notebook completed: create managed table in silver schema")
dbutils.notebook.exit(json.dumps({"status": "SUCCESS", "notebook": "create_managed_table_silver"}))
```
