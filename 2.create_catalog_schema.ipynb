```python
# Create Unity Catalog & Schemas

Provisions the **Unity Catalog** (`finance_dev`) and its three layer-specific **schemas**
(`bronze_dev`, `silver_dev`, `gold_dev`), each pinned to its own ADLS Gen2 container.

> All names and paths come from `0.config` — no hardcoded values here.
```

```python
# ── Step 0: Load shared configuration ────────────────────────────────────────
%run ./0.config
```

```python
# ── Step 1: Create Catalog and Schemas ───────────────────────────────────────

spark.sql(f"""
CREATE CATALOG IF NOT EXISTS {CATALOG_NAME}
MANAGED LOCATION '{CATALOG_ROOT}'
""")

spark.sql(f"USE CATALOG {CATALOG_NAME}")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {BRONZE_SCHEMA}
MANAGED LOCATION '{BRONZE_PATH}'
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {SILVER_SCHEMA}
MANAGED LOCATION '{SILVER_PATH}'
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {GOLD_SCHEMA}
MANAGED LOCATION '{GOLD_PATH}'
""")
```

```python
# ── Step 2: Validate catalog metadata ────────────────────────────────────────

spark.sql(f"DESCRIBE CATALOG EXTENDED {CATALOG_NAME}").display()

for schema in [BRONZE_SCHEMA, SILVER_SCHEMA, GOLD_SCHEMA]:
    print(f"\n── {CATALOG_NAME}.{schema} ──")
    spark.sql(f"DESCRIBE SCHEMA EXTENDED {CATALOG_NAME}.{schema}").display()
```
