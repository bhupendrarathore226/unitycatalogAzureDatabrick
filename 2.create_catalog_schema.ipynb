```python
import logging
import json

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s - %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)
logger = logging.getLogger("create_catalog_schema")
logger.info("Notebook started: create_catalog_schema")
logger.info("Starting: Create catalog finance_dev and schemas bronze_dev, silver_dev, gold_dev")
```

```python
try:
    spark.sql("""
        CREATE CATALOG IF NOT EXISTS finance_dev
        MANAGED LOCATION 'abfss://finance-dev@dbstoragefinance.dfs.core.windows.net/'
    """)
    spark.sql("USE CATALOG finance_dev")
    spark.sql("""
        CREATE SCHEMA IF NOT EXISTS bronze_dev
        MANAGED LOCATION 'abfss://bronze@dbstoragefinance.dfs.core.windows.net/'
    """)
    spark.sql("""
        CREATE SCHEMA IF NOT EXISTS silver_dev
        MANAGED LOCATION 'abfss://silver@dbstoragefinance.dfs.core.windows.net/'
    """)
    spark.sql("""
        CREATE SCHEMA IF NOT EXISTS gold_dev
        MANAGED LOCATION 'abfss://gold@dbstoragefinance.dfs.core.windows.net/'
    """)
    logger.info("Completed: Catalog finance_dev and schemas bronze_dev, silver_dev, gold_dev created successfully")
except Exception as e:
    logger.error("Failed to create catalog or schemas. Error: %s", str(e), exc_info=True)
    dbutils.notebook.exit(json.dumps({"status": "FAILED", "step": "create_catalog_schema", "error": str(e)}))
    raise
```

```python
logger.info("Starting: Describe catalog finance_dev")
```

```python
try:
    spark.sql("DESCRIBE CATALOG EXTENDED finance_dev").display()
    logger.info("Completed: Catalog finance_dev described successfully")
except Exception as e:
    logger.error("Failed to describe catalog finance_dev. Error: %s", str(e), exc_info=True)
    dbutils.notebook.exit(json.dumps({"status": "FAILED", "step": "describe_catalog_finance_dev", "error": str(e)}))
    raise
```

```python
logger.info("Starting: Query sample data from finance_dev.bronze_dev.results")
```

```python
logger.info("Notebook completed: create_catalog_schema")
dbutils.notebook.exit(json.dumps({"status": "SUCCESS", "notebook": "create_catalog_schema"}))
```
